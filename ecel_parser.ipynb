# Проход по ссылкам внутри первоначального документа и скачка всех файлов по ним

```python
def download_file_stream(url, local_filename):
    with requests.get(url, stream=True) as r:
        r.raise_for_status()
        with open(local_filename, 'wb') as f:
            for chunk in r.iter_content(chunk_size=8192):
                f.write(chunk)
    print(f'Файл успешно сохранён как {local_filename}')
```

```python
def download_files(source_doc):
    """
    Ищет ссылки внутри source (представляющий собой список из json) и скачивает файлы
    source_doc
    """
    doc_number = 0
    for obj in source_doc:
        doc_number += 1
        link_number = 0
        for k, v in obj.items():
            if v.startswith('http'): # определяем ссылки
                link_number += 1
                base = v.split('/edit')[0]
                link_for_download = base + "/export?format=xlsx"
                local_filename = f'./downloaded_all/{doc_number}_{link_number}.xlsx'
                download_file_stream(url=link_for_download, local_filename=local_filename)

                
    
```

# Обработка скачанных excel файлов
Оставляем только те, в которых есть поля:
1. "Разработка материалов утверждаемой части"
2. "Согласование утв. части в АИС СД, утверждение ППМ"

```python
values_to_check = ["Разработка материалов утверждаемой части", "Согласование утв. части в АИС СД, утверждение ППМ"]
def check_values(df):
    value_1_found = False
    value_2_found = False
    for index, row in df.iterrows():
        for cell_value in row.values:
            cell_value = str(cell_value)
            if values_to_check[0] in cell_value:
                value_1_found = True
            if values_to_check[1] in cell_value:
                value_2_found = True
            # if (values_to_check[0] in cell_value) or (values_to_check[1] in cell_value):
            # if values_to_check[0] in row.values:
                # print(row.values)
            # if values_to_check[1] in row.values:
            #     print(f"Найдено! в строке {index}")
            #     pprint(df.iloc[index:])
            if value_1_found or value_2_found:
                return True
    return False
```

```python
for file in Path('./downloaded_all').iterdir():
    xl = pd.ExcelFile(file)
    sheet_names = xl.sheet_names
    for sheet in sheet_names:
        df = xl.parse(sheet)
        if check_values(df):
            print(f"file = {file} , sheet_name = {sheet}")
        # print(df)
```
