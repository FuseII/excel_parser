# Проход по ссылкам

```python
def download_file_stream(url, local_filename):
    """
    Скачивает файл по ссылке
    :param url: ссылка
    :param local_filename: имя сохраняемого файла
    :return: 
    """
    with requests.get(url, stream=True) as r:
        r.raise_for_status()
        with open(local_filename, 'wb') as f:
            for chunk in r.iter_content(chunk_size=8192):
                f.write(chunk)
    print(f'Файл успешно сохранён как {local_filename}')
```

```python
def parse_excel(local_filename):
    """Парсинг excel файла в json"""
    inner_df = pd.ExcelFile(local_filename)
    inner_df = inner_df.parse(inner_df.sheet_names[0])
    inner_df = inner_df.astype(str)
    inner_df.dropna(axis=0, how='all', inplace=True)
    inner_df = inner_df.iloc[3:]
    inner_df.columns = inner_df.iloc[0]
    inner_df = inner_df.drop(3).reset_index(drop=True)
    inner_df.columns.values[3] = 'Дата'
    inner_df.columns.values[-1] = 'Статус'
    inner_df.columns = inner_df.columns.str.replace('\n', '', regex=False)
    inner_df.columns = inner_df.columns.str.strip()
    inner_df = inner_df.replace(r'\n',' ', regex=True)
    inner_df = inner_df.replace(r'\t',' ', regex=True)
    result_list = []
    i = 0
    while i < len(inner_df):
        result = dict()
        if (inner_df["Наименование работы"][i] != 'nan') and (inner_df["Наименование работы"][i+1] == 'nan'):
            step = 2
        else:
            step = 1
        part_df = inner_df.iloc[i:i+step]  # Получаем одну/две строки
        result = part_df.iloc[0,:3].to_dict() # попробовать dict
        result.update(part_df[["Статус"]].iloc[[0]].to_dict(orient='records')[0])
        i+=step
        part_df = part_df.set_index('Дата').iloc[:,3:-1].to_dict()
        result.update(part_df)
        result_list.append(result)

    return result_list
    
    
    
```

```python
columns_to_be_matched = ['Разработка материалов существующего положения + эскизы', 'Рассмотрение эскизов, выбор рекомендуемого варианта', 'Разработка обосновывающих материалов проектного предложения', 'Расмотрение обосновывающих материалов, получение замечаний', 'Корректировка обосновывающих материалов по замечаниям ОИВ, повторная рассылка', 'Корректировка материалов по замечаниям ОИВ, Подготовка и согласование макета и НПМ, получение согласований ОИВ', 'Подготовка презентации на ГЗК', 'Проведение РГ ГЗК И ГЗК', 'Подготовка Утв . Части', 'ПУ МКА', 'АИС СД УКД', 'АИС СД ОИВ', 'Доработка материалов по замч ОИВ', 'ПУ ПМ', 'ОАУ ПМ']
def check_excel(local_filename):
    """
    Прооверяет что это подходящий excel документ
    """
    df = pd.ExcelFile(local_filename)
    df = df.parse(df.sheet_names[0])
    df = df.astype(str)
    df.dropna(axis=0, how='all', inplace=True)
    df = df.iloc[3:]
    df.columns = df.iloc[0]
    df.columns = df.columns.str.replace('\n', '', regex=False)
    df.columns = df.columns.str.strip()
    df = df.replace(r'\n',' ', regex=True)
    df = df.replace(r'\t',' ', regex=True)
    df = df.drop(3).reset_index(drop=True)
    matched_columns = 0
    for column in columns_to_be_matched:
        if column in df.columns:
            matched_columns += 1
    if matched_columns == len(columns_to_be_matched):
        return True
    else:
        return False
           
    
```

```python
# множество посещённых ссылок
visited_links = []
```

```python
def follow_links(source_doc):
    """
    Ищет ссылки внутри source (представляющий собой список из json) и скачивает файлы
    source_doc
    """
    json_number = 0
    for obj in source_doc:
        json_number+=1
        obj["links"] = dict()
        for k, v in obj.items():
            if isinstance(v, str) and v.startswith('http'): # определяем ссылки
                
                base = v.split('/edit')[0]
                link_for_download = base + "/export?format=xlsx"
                
                if link_for_download not in visited_links:
                    visited_links.append(link_for_download)
                    link_number = visited_links.index(link_for_download)
                    local_filename = f'./downloaded_all/{link_number}.xlsx'
                    download_file_stream(url=link_for_download, local_filename=local_filename)
                else:
                    link_number = visited_links.index(link_for_download)
                    local_filename = f'./downloaded_all/{link_number}.xlsx'
                    
                if check_excel(local_filename):
                    print(f"{local_filename}")
                    parsed_data = parse_excel(local_filename)
                    obj["links"].update({k:parsed_data})
                    with open(f'./saved_jsons/{json_number}.json', 'w',encoding='utf-8') as f:
                        json.dump(obj, f, ensure_ascii=False, indent=4)
                # фильтровать excel файлы
```
